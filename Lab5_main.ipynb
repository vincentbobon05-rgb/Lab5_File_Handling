```python
from pathlib import Path
from datetime import datetime
import shutil

backup_dir = Path.home() / "Documents" / "Surname_Activity_5"
backup_dir.mkdir(exist_ok=True)

def write_with_backup(filename: str, content: str):
    file_path = backup_dir / filename

    # Create backup if file exists
    if file_path.exists():
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        backup_path = file_path.with_name(
            f"{file_path.stem}_Ilagan_backup_{timestamp}{file_path.suffix}"
        )
        shutil.copy2(file_path, backup_path)
        print(f"Backup saved: {backup_path.name}")

    # Choose overwrite or append
    mode = input("Enter mode (w = overwrite, a = append): ")
    with open(file_path, mode, encoding="utf-8") as file:
        file.write(content)

    print(f"File saved: {file_path.name}")

# List backups for a specific file
def list_backups(filename):
    for file in backup_dir.glob(f"{Path(filename).stem}*backup*"):
        print(file.name)

# Demo
write_with_backup("demo.txt", "Hello World")
list_backups("demo.txt")
```

```python
from pathlib import Path
from datetime import datetime
import shutil

backup_dir = Path.home() / "Documents" / "Ilagan_Activity_5"
backup_dir.mkdir(exist_ok=True)

def file_manager():
    file_name = input("Enter filename (e.g., notes.txt): ")
    file_path = backup_dir / file_name

    while True:
        print("\n--- MENU ---")
        print("1. Write to file")
        print("2. Append to file")
        print("3. Read file")
        print("4. Backup file")
        print("5. Exit")
        print("6. List backups")

        choice = input("Choose an option (1-6): ")

        if choice == "1":
            content = input("Enter content:\n")
            with open(file_path, "w", encoding="utf-8") as f:
                f.write(content)
            print("File written successfully.")

        elif choice == "2":
            more = input("Append content:\n")
            with open(file_path, "a", encoding="utf-8") as f:
                f.write("\n" + more)
            print("Content appended.")

        elif choice == "3":
            if file_path.exists():
                with open(file_path, "r", encoding="utf-8") as f:
                    print("\nFile Content:\n", f.read())
            else:
                print("File not found.")

        elif choice == "4":
            if file_path.exists():
                timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
                backup = file_path.with_name(
                    f"{file_path.stem}_Ilagan_backup_{timestamp}{file_path.suffix}"
                )
                shutil.copy2(file_path, backup)
                print("Backup created:", backup.name)
            else:
                print("File not found.")

        elif choice == "5":
            print("Exiting...")
            break

        elif choice == "6":
            for file in backup_dir.glob("*backup*"):
                print(file.name)

        else:
            print("Invalid choice.")

if __name__ == "__main__":
    file_manager()
```

```python
# Merge Two Files
from pathlib import Path

# Personal details
student_id = "TUPM-25-3550"

# Define file paths
documents_path = Path.home() / "Documents" / "Ilagan_Activity_5"
f1 = documents_path / f"intro_{student_id}.txt"
f2 = documents_path / f"lines_{student_id}.txt"
merged = documents_path / f"merged_{student_id}.txt"

# Ensure both source files exist
if not f1.exists():
    f1.write_text(f"Welcome {student_id} to File Handling in Python!")
    print(f"Sample intro file created for Student ID: {student_id}")

if not f2.exists():
    f2.write_text("Line 1\nLine 2\nLine 3")
    print(f"Sample lines file created for Student ID: {student_id}")

# Merge contents of both files
with merged.open("w") as mf:
    mf.write(f1.read_text())
    mf.write("\n")
    mf.write(f2.read_text())

print(f"Files merged successfully for Student ID: {student_id}")
```

```python
from pathlib import Path

# folder path
folder = Path.home() / "Documents" / "Ilagan_Activity_5"

files = list(folder.glob("*.txt"))

results = []

for file in files:
    text = file.read_text()
    lines = [line for line in text.splitlines() if line.strip() != ""]

    num_lines = len(lines)
    num_words = len(text.split())
    num_chars = len(text.replace(" ", ""))

    if num_lines > 0 and num_words > 0:
        words_per_line = num_words / num_lines
        chars_per_word = num_chars / num_words
    else:
        words_per_line = 0
        chars_per_word = 0

    results.append((file.name, num_lines, num_words, num_chars, words_per_line, chars_per_word))

# sort by words per line (descending)
results.sort(key=lambda x: x[4], reverse=True)

# print table
print("Filename | Lines | Words | Chars | Words/Line | Chars/Word")
for r in results:
    print(f"{r[0]} | {r[1]} | {r[2]} | {r[3]} | {round(r[4],2)} | {round(r[5],2)}")

# show most and least dense
if results:
    print("\nMost dense file:", results[0][0])
    print("Least dense file:", results[-1][0])
```
